# Probability Theory
### Conditional Probability : 
* Probability of an event A given that event B has occured,    $ P(A \mid B) = \frac{P(A \mid B)}{P(B)} $

#### Bayes theorem :          $ P(A \mid B) = \frac{P(B \mid A)\,P(A)}{P(B)} $

## Common Probability Distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom,poisson,norm
```

#### Gaussian Distribution (Normal disrtibution):
* Bell shaped curve with mean and standard deviation    $$f(x) =  \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{x-\mu)^2}{2\sigma^2}} $$

```python
mu,sigma = 0,1
x = np.linspace(-4,4,100)
y = norm.pdf(x,mu,sigma)
plt.plot(x,y,label = "Gaussian")
plt.title("Gaussian distributuion")
plt.show()
```

#### Binomial Distribution :
* models the number of success in n independent bernouli trials.
* Application in ML : logistic regression to model binary outcomes


```python
n,p = 10,0.5
x = np.arange(0,n+1)
y = binom.pmf(x,n,p)
plt.bar(x,y,color = "blue")
plt.title("Binomial distributuion")
plt.show()
```

#### Bernoulli Distribution
* describe the outcomes of binary experiment(success or failure)
* P(X=1) = P ---> success
* P(X=0) = 1-P ---> failure

```python
p = 0.6
plt.bar([0,1],[1-p,p],color = "blue")
plt.title("Bernoulli distributuion")
plt.xticks([0,1],labels = ["0(failure)","1(success)"])
plt.show()
```

#### Poisson Distribution 
* models the number of events in a fixed intervals of time or space
* $ f(x) = \frac{\lambda^k e^{-\lambda}}{k!} $

```python
lam = 3
x = np.arange(0,10)
y = poisson.pmf(x,lam)
plt.bar(x,y,color = "pink")
plt.title("Poisson distributuion")
plt.show()
```
