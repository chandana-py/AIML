```python
#features of dispersion
''' tells us how spread out thae data values are form the center (mean/median)'''
```

```python
#Range:
'''Difference '''
```

```python
data = [10,20,30,40,50]
range_value = max(data)-min(data)
print("Range:",range_value)
```

```python
#variance
'''Average of squared difference of the mean'''
```

```python
#standard deviation
'''sqaure root of variance
low standard deviation-------------data is close mean
high standard deviation--------------data is widely spread'''
```

```python
import numpy as np
data =[10,20,30]
print("variance = ",np.var(data))
```

```python
print("std:",np.std(data))
```

```python
#Quartiles
'''Quartiles divide a sorted dataset into 4 equal parts'''
```

```python
#main quartiles
'''
Quartiles        Meaning                     Position
Q1                First quartile            25% data below
Q2                second quartile            50% 
Q3                third quartile             75%
'''          
```

```python
data = [10,20,30,40,50,60,70,80]
q1 = np.percentile(data ,25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data ,75)
print("Q1:",q1)
print("Q2:",q2)
print("Q3:",q3)
```

```python
print("IQR:",q3-q1)
```

```python
'''
Measure               formula                            use                    limitation
range                 Max-min                          Quick spread             sensitive to outliers
variance           avg squared difference        overall spread                hard to interpret
std:dev           sq.root of variance              most useful                   still affected by outliers 
IQR                Q3-Q2                         robust spread                  Ignores outliers'''
```
