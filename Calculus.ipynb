## Derivatives
It measures rate at which function changes with respect to its input.
* For function f(x) , it denotes as f'(x)
* It indicates the slope of tangent line at a point

```python
import sympy as sp
```

#### 1.  f(x) = x<sup>2</sup>

```python
x = sp.Symbol('X')
f = x**2
der = sp.diff(f,x)
print("derivative of f: \n" , der)
```

## Partial derivatives
It measures how a function changes with respect to one variable while keeping other variable constant
* For function , f(x,y) = x<sup>2</sup>+y<sup>2</sup> ,

                            df/dx = 2x and df/dy = 2y

## Gradients 
Vector of all Partial derivatives, indicated directions of the steepest ascent
* For function , f(x,y) = x<sup>2</sup>+y<sup>2</sup>,

                                grad(f) = [df/dx , df/dy] = [2x,2y]

```python
import sympy as sp
x,y = sp.symbols('x  y')
f = x**2+y**2
grad_x = sp.diff(f,x)
grad_y = sp.diff(f,y)
print("Partial derivatives: ",[grad_x,grad_y])
```

```python
x,y = sp.symbols('x  y')
f = x**2+3*y**2-4*x*y
grad_x = sp.diff(f,x)
grad_y = sp.diff(f,y)
print("grad X: ",grad_x)
print("grad Y: ",grad_y)
```

### Gradient descent for Linear Regression

```python
import numpy as np
#define Gradient descent function
def gradient_descent(x,y,theta,learning_rate,iterations):
    m=len(y)
    for _ in range(iterations):
        predictions = np.dot(x,theta)
        errors = predictions-y
        gradients = (1/m)*np.dot(x.T,errors)
        theta -= learning_rate*gradients
    return theta
x = np.array([[1,1],[1,2],[1,3]])
y = np.array([2,2.5,3.5])
theta = np.array([0.1,0.1])
learning_rate = 0.1
iterations = 1000
optimized_theta = gradient_descent(x,y,theta,learning_rate,iterations)
print(optimized_theta)
```

## Integrals

```python
import sympy as sp
```

#### 1. f(x) = x<sup>2</sup>

```python
x = sp.Symbol('X')
f = x**2
definite_integral = sp.integrate(f,(x,0,2))
print("definite_integral: \n ",definite_integral)
indefinite_integral = sp.integrate(f,(x))
print("\n indefinite_integral: \n ",indefinite_integral)
```

#### 2. f(x) = e<sup>-x</sup>

```python
x = sp.Symbol('X')
f = sp.exp(-x)
definite_integral = sp.integrate(f,(x,0,sp.oo))
print("definite_integral: \n ",definite_integral)
indefinite_integral = sp.integrate(f,(x))
print("\n indefinite_integral: \n ",indefinite_integral)
```

#### 3. f(x) = -5x+(2/3)y

```python
x,y = sp.symbols('X Y')
f = -5*x+(2/3)*y
definite_integral = sp.integrate(f,(x,0,2))
print("definite_integral: \n ",definite_integral)
indefinite_integral = sp.integrate(f,(x))
print("\n indefinite_integral: \n ",indefinite_integral)
```

## Stochastic Gradient Descent (SGD)
Optimization algorithm that uses random subsets of data to compute gradients and update parameters.
###### Its varients are :
* mini batch SGD
* momentum
* Adam optimizer
   

```python
# compute SGD on linear model
```

```python
import numpy as np
np.random.seed(42)

# Generate synthetic data
x = 2* np.random.rand(100,1)
y = 4+3*x+np.random.rand(100,1)
x_b = np.c_[np.ones((100,1)),x]

# SGD implementation
def stochastic_gradient_descent(x,y,theta,learning_rate,n_epochs):
    m = len(y)
    for epoch in range(n_epochs):
        for i in range(m):
            random_index = np.random.randint(m)
            xi = x[random_index:random_index+1]
            yi = y[random_index:random_index+1]
            gradients = 2* xi.T @ (xi @ theta -yi)
            theta -= learning_rate*gradients
    return theta
# iniatilaize parameters
theta = np.random.randn(2,1)
learning_rate = 0.01
n_epochs = 50

# perform SGD

theta_optimized = stochastic_gradient_descent(x_b,y,theta,learning_rate,n_epochs)
print("Optimized_parameters: \n ",theta_optimized)
```
